# Study Case 01 — Data Quality & Reconciliation

Python only runs SQL and displays the database-side QA results.


```python
import os
import psycopg
from IPython.display import display
conn = psycopg.connect(os.environ['SUPABASE_DB_URL'])
```

```python
sql = "SELECT * FROM analytics.v_acnc_validation_summary;"
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```

```python
sql = """
SELECT check_group, check_name, source_column, source_rows, database_rows,
       source_non_nulls, database_non_nulls, source_nulls, database_nulls,
       source_value, database_value, difference, difference_pct, threshold, status
FROM analytics.acnc_data_quality_report
ORDER BY CASE status WHEN 'FAIL' THEN 1 WHEN 'PASS' THEN 2 ELSE 3 END, check_group, check_name, source_column;
"""
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```
