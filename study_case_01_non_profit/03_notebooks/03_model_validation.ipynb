# Study Case 01 — Data Model Validation

Validate the relational model using SQL. Python is only the execution/display layer.


```python
import os, psycopg
from IPython.display import display
conn = psycopg.connect(os.environ['SUPABASE_DB_URL'])
```

```python
sql = """
SELECT 'charities' table_name, COUNT(*) row_count FROM acnc.charities
UNION ALL SELECT 'operations', COUNT(*) FROM acnc.operations
UNION ALL SELECT 'workforce', COUNT(*) FROM acnc.workforce
UNION ALL SELECT 'financials', COUNT(*) FROM acnc.financials
UNION ALL SELECT 'reporting', COUNT(*) FROM acnc.reporting
UNION ALL SELECT 'registration', COUNT(*) FROM acnc.registration
UNION ALL SELECT 'fundraising', COUNT(*) FROM acnc.fundraising;
"""
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```

```python
sql = """
SELECT COUNT(*) AS orphan_financials
FROM acnc.financials f
LEFT JOIN acnc.charities c ON c.charity_id=f.charity_id
WHERE c.charity_id IS NULL;
"""
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```
