# Study Case 01 — ACNC Source Profile

Python is used only to execute and display SQL results.
All profiling logic remains in PostgreSQL.


```python
import os
import psycopg
from IPython.display import display
conn = psycopg.connect(os.environ['SUPABASE_DB_URL'])
```

```python
sql = "SELECT * FROM analytics.v_acnc_source_profile;"
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```

```python
sql = """
SELECT table_schema, table_name, COUNT(*) AS columns
FROM information_schema.columns
WHERE table_schema IN ('staging','acnc')
GROUP BY 1,2
ORDER BY 1,2;
"""
with conn.cursor() as cur:
    cur.execute(sql)
    display(cur.fetchall())
```
